## Завдання 1. Створення масивів

**Київ, варіант 1.** Температури взято з таблиці завдання (12 місяців від січня до грудня). Кількість днів відповідає невисокосному календарному року. Масив `month_numbers` створено за допомогою `np.arange(1, 13)`.

````
city = "Київ"
variant = 1
months = np.array(["Січень", "Лютий", "Березень", "Квітень", "Травень", "Червень",
                   "Липень", "Серпень", "Вересень", "Жовтень", "Листопад", "Грудень"])

temps = np.array([-4, -3, 2, 9, 15, 19, 21, 20, 14, 8, 2, -2])
days_in_month = np.array([31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31])
month_numbers = np.arange(1, 13)

for name, arr in [("temps", temps), ("days_in_month", days_in_month),
                  ("month_numbers", month_numbers)]:
    print(f"{name}: {arr}")
    print(f"  shape: {arr.shape}, dtype: {arr.dtype}")
````

## Завдання 2. Векторизовані обчислення

Формула переведення температури: **°F = °C × 9/5 + 32**.

Для підрахунку теплих днів фільтруються дні тих місяців, де середня температура строго перевищує 15 °C (`temps > 15`). Травень має рівно 15 °C, тому не включається.

````
temps_fahrenheit = temps * 9 / 5 + 32
warm_mask = temps > 15
warm_days = days_in_month[warm_mask].sum()
total_days = days_in_month.sum()

print("Температури у °F:", temps_fahrenheit)
print("Місяці з температурою > 15 °C:", months[warm_mask])
print("Кількість днів у цих місяцях:", days_in_month[warm_mask])
print("Загалом днів із середньомісячною температурою > 15 °C:", warm_days)
print("Кількість днів за рік:", total_days)
assert total_days == 365, "Очікується 365 днів для невисокосного року"
print("Перевірка: 365 днів — успішно.")
````

**Результат виконання:**


## Завдання 3. Індексація і фільтрація

Методи `argmax()` та `argmin()` повертають індекси елементів (відлік від 0). Порядковий номер місяця береться з `month_numbers`.

````
max_idx = temps.argmax()
min_idx = temps.argmin()
negative_month_numbers = month_numbers[temps < 0]
sorted_temps = np.sort(temps)

print(f"Найтепліший місяць: {months[max_idx]} (№{month_numbers[max_idx]}), {temps[max_idx]} °C; індекс {max_idx}")
print(f"Найхолодніший місяць: {months[min_idx]} (№{month_numbers[min_idx]}), {temps[min_idx]} °C; індекс {min_idx}")
print("Номери місяців із температурою нижче 0 °C:", negative_month_numbers)
print("Відсортовані температури:", sorted_temps)
print("Оригінальний масив temps (не змінено):", temps)
````

## Завдання 4. Двовимірний масив та `axis`

Додаємо умовні години сонячного сяйва для Києва. Отримуємо матрицю розмірності `(3, 12)`:

* рядок 0 — температура (°C);
* рядок 1 — кількість днів у місяці;
* рядок 2 — умовні сонячні години.

````
sunshine_hours = np.array([40, 60, 110, 160, 220, 250, 270, 240, 160, 100, 50, 30])
data = np.array([temps, days_in_month, sunshine_hours])

row_sums = data.sum(axis=1)
row_means = data.mean(axis=1)
column_sums = data.sum(axis=0)

print("Масив data (3 × 12):")
print(data)
print("Форма масиву:", data.shape)
print("Сума по кожному рядку (axis=1):", row_sums)
print("Середнє по кожному рядку (axis=1):", row_means)
print("Суми по кожному стовпцю (axis=0):", column_sums)
print("Січень:", temps[0], "+", days_in_month[0], "+", sunshine_hours[0], "=", column_sums[0])
assert data.shape == (3, 12)
````

**Результат виконання:**

```text
Масив data (3 × 12):
[[ -4  -3   2   9  15  19  21  20  14   8   2  -2]
 [ 31  28  31  30  31  30  31  31  30  31  30  31]
 [ 40  60 110 160 220 250 270 240 160 100  50  30]]
Форма масиву: (3, 12)
Сума по кожному рядку (axis=1): [  99  365 1690]
Середнє по кожному рядку (axis=1): [  8.25        30.41666667 140.83333333]

```

**Пояснення для першого стовпця (січень):**
Результат дорівнює **67** (-4 °C + 31 день + 40 сонячних годин). Число є суто формальним результатом математичної операції додавання елементів стовпця матриці і не несе фізичного змісту, оскільки об'єднує різнорідні показники.

Параметр `axis=1` виконує операцію вздовж стовпців для кожного рядка окремо (довжина результату 3), а `axis=0` — вздовж рядків для кожного стовпця (довжина результату 12).

## Завдання 5. `reshape` і broadcasting

Спроба додати `rows` форми `(2, 12)` та одновимірний масив `correction` форми `(2,)` призводить до помилки `ValueError`. Правило broadcasting в NumPy вимагає порівняння осей **справа наліво**: оскільки крайні розміри `12` і `2` не рівні між собою і жоден із них не дорівнює `1`, операція неможлива.

Демонстрація очікуваної помилки:

````
rows = np.array([temps, days_in_month])
correction = np.array([1, 0])
print("Форма rows:", rows.shape)
print("Форма correction:", correction.shape)

try:
    invalid_result = rows + correction
except ValueError as exc:
    print("Очікувана помилка ValueError:", exc)
else:
    raise AssertionError("Очікувалася помилка broadcasting")
````

**Результат виконання:**

```text
Форма rows: (2, 12)
Форма correction: (2,)
Очікувана помилка ValueError: operands could not be broadcast together with shapes (2,12) (2,)

```

Виправлення форми через `.reshape(2, 1)`:

````
correction_2d = correction.reshape(2, 1)
corrected_rows = rows + correction_2d

print("Форма correction після reshape:", correction_2d.shape)
print("Форма результату:", corrected_rows.shape)
print("Масив після додавання поправки до кожного рядка:")
print(corrected_rows)

assert corrected_rows.shape == (2, 12)
assert np.array_equal(corrected_rows[0], temps + 1)
assert np.array_equal(corrected_rows[1], days_in_month)
print("Перевірка: +1 до температур, +0 до кількості днів — успішно.")
````

**Результат виконання:**

```text
Форма correction після reshape: (2, 1)
Форма результату: (2, 12)
Масив після додавання поправки до кожного рядка:
[[ -3  -2   3  10  16  20  22  21  15   9   3  -1]
 [ 31  28  31  30  31  30  31  31  30  31  30  31]]
Перевірка: +1 до температур, +0 до кількості днів — успішно.

```

**Пояснення:** Форми `(2, 12)` і `(2, 1)` є повністю сумісними. При порівнянні справа наліво розмір `1` розтягується (broadcast) до `12`, а перший вимір `2` повністю збігається. Таким чином значення `1` додалося до кожного елемента температури, а `0` — до кількості днів.

## Письмові відповіді на контрольні запитання

1. **Яка різниця між `shape == (12,)` та `shape == (1, 12)`?**
`(12,)` — це одновимірний масив (вектор) із 12 елементами й однією віссю. `(1, 12)` — це двовимірний масив (матриця) з двома осями, що містить 1 рядок і 12 стовпців. Різниця визначає їхню поведінку під час індексації, транспонування та застосування правил broadcasting.
2. **Чому `temps[temps > 15]` повертає масив іншої довжини?**
Булева індексація працює за принципом фільтра. Вираз `temps > 15` створює булевий масив значень `True` та `False`. У підсумковий масив потрапляють лише ті елементи, де умова виявилася істинною. Для Києва таких місяців лише три (`[19, 21, 20]`), тому довжина масиву становить 3 замість 12.
3. **Що станеться, якщо додати масив форми `(3,)` до масиву форми `(3, 12)`?**
Виникне виняток **`ValueError`**. За правилами broadcasting зіставлення вимірів виконується справа наліво: порівнюються `12` і `3`. Вони не рівні між собою і жоден не дорівнює `1`, тому розтягнення неможливе. Щоб операція пройшла успішно, вектор форми `(3,)` потрібно перетворити на матрицю-стовпець форми `(3, 1)`.
4. **Чому `rows + correction` дає помилку, а `rows + correction.reshape(2, 1)` — ні?**
Форми `(2, 12)` та `(2,)` конфліктують по останньому виміру (`12 ≠ 2`). Після виконання `correction.reshape(2, 1)` форми стають `(2, 12)` та `(2, 1)`. Тепер виміри сумісні: по другій осі значення `1` дублюється 12 разів, а по першій виміри `2` і `2` збігаються. Поправка `+1` коректно застосовується до першого рядка, а `+0` — до другого.
