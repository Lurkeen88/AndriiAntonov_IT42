## Крок 1. Побудова DataFrame свого варіанта

Дані для Києва: середньомісячна температура (`avg_temp` з Практики 1) та кількість опадів (`precip` з поточної таблиці).

````
import pandas as pd

months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
          "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]
avg_temp = [-4, -3, 2, 9, 15, 19, 21, 20, 14, 8, 2, -2]
precip = [38, 36, 38, 41, 53, 74, 84, 69, 47, 36, 44, 44]

df = pd.DataFrame({
    "month": months,
    "avg_temp": avg_temp,
    "precip": precip,
})

print("Перші рядки DataFrame:")
print(df.head())
print("\nФорма (shape):", df.shape)
print("\nТипи даних (dtypes):\n", df.dtypes)
````

## Завдання 2. Відбір рядків і стовпців

Зміна індексу на номери місяців (від 1 до 12) та демонстрація різниці між `.loc` (за міткою) і `.iloc` (за позицією).

## Завдання 3. Фільтрація за умовою

Застосування логічних масок для відбору рядків. Використовуються побітові оператори `&` та `|`.

````
print("1. Місяці, де avg_temp > 15:\n", df[df["avg_temp"] > 15])

print("\n2. Місяці, де avg_temp > 10 і precip > 50 (&):\n", df[(df["avg_temp"] > 10) & (df["precip"] > 50)])

print("\n3. Місяці, де avg_temp < 0 або precip > 80 (|):\n", df[(df["avg_temp"] < 0) | (df["precip"] > 80)])
````

## Завдання 4. Обчислювані стовпці

Створення нових стовпців шляхом векторизованої арифметики та застосування власної функції через `.apply()`.

````
# 1. Стовпець avg_temp_f (векторизована формула)
df["avg_temp_f"] = df["avg_temp"] * 9/5 + 32

# 2. Булевий стовпець is_wet (перевищення середньої кількості опадів)
mean_precip = df["precip"].mean()
print(f"Середня кількість опадів по таблиці: {mean_precip:.2f} мм\n")
df["is_wet"] = df["precip"] > mean_precip

# 3. Текстовий стовпець season через .apply()
def get_season(month_num):
    if month_num in [12, 1, 2]:
        return "зима"
    elif month_num in [3, 4, 5]:
        return "весна"
    elif month_num in [6, 7, 8]:
        return "літо"
    else:
        return "осінь"

# Оскільки індекс таблиці встановлено від 1 до 12, функцію можна застосувати до нього
df["season"] = df.index.to_series().apply(get_season)

print("Фінальний DataFrame:\n", df)
````

## Письмові відповіді на контрольні запитання

1. **Чи можете пояснити, чому `.loc[1:6]` і `.iloc[0:6]` можуть повернути різну кількість рядків, залежно від того, який саме індекс у таблиці?**
* Метод `.loc[1:6]` шукає рядки за їхніми **підписами (мітками)**. Обидві межі зрізу (і 1, і 6) включаються у вибірку. Якщо індекс є текстовим, він шукатиме конкретні слова, а якщо числовий не по порядку — витягне всі рядки, що знаходяться між міткою `1` та міткою `6`.
* Метод `.iloc[0:6]` звертається до **фізичних позицій** рядків у пам'яті (відлік від нуля). Верхня межа `6` не включається, тому він жорстко поверне рівно 6 перших рядків (з 0-го по 5-й), незалежно від того, які мітки присвоєні цим рядкам в індексі.


2. **Чи розумієте, чому `df["avg_temp"] > 10 and df["precip"] > 50` викличе помилку, а `(df["avg_temp"] > 10) & (df["precip"] > 50)` — ні?**
Стандартний оператор Python `and` очікує отримати єдине значення `True` або `False` для всього виразу, але об'єкт `Series` складається з багатьох значень, що спричиняє помилку `ValueError` (неоднозначність). У Pandas логічне множення виконується **поелементно**, для цього визначений побітовий оператор `&`. Дужки обов'язкові через те, що `&` має вищий пріоритет за оператори порівняння.
3. **Чи можете описати словами, у чому різниця між обчислюваним стовпцем, створеним формулою, і створеним через `.apply()`?**
* **Формула:** Використовує механізм векторизації (під капотом NumPy), застосовуючи одну математичну операцію одночасно до всього стовпця. Це працює максимально швидко.
* **Через `.apply()`:** Застосовує вказану Python-функцію до кожного елемента стовпця по черзі, імітуючи цикл `for`. Це працює повільніше за векторизацію, але є необхідним інструментом, коли логіка містить складні умови (як визначення сезонів) або нетипову обробку тексту.
