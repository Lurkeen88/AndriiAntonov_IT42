## Завдання 1. Класифікація змінних за шкалою вимірювання

У нашому наборі є три змінні: **місто**, **номер місяця** (1–12) та **кількість опадів (мм)**.

1. **Визначення шкали та обґрунтування:**
* **Місто:** *Номінальна шкала*. Значення є категоріями (назви населених пунктів) без жодного кількісного ранжування чи порядку.
* **Номер місяця (1–12):** Розглядається нижче окремо.
* **Кількість опадів (мм):** *Шкала відношень*. Значення мають абсолютний нуль (0 мм означає повну відсутність опадів), а відношення значень мають фізичний сенс (наприклад, 80 мм опадів у два рази більші за 40 мм).


2. **Особливий розгляд «номера місяця»:**
Місяці утворюють чітку хронологічну послідовність (порядкова шкала), а інтервали між ними календарно рівні (місяць триває приблизно рівні проміжки часу). Проте у цієї шкали **немає природного нуля** (не існує поняття «нульового місяця» або «відсутності місяця як початку відліку»). Через це «номер місяця» є класичним прикладом **порядкової шкали з рівними інтервалами** (або межовим випадком між порядковою та інтервальною), де додавання чи множення номерів місяців не має змісту.
3. **Коректні міри положення:**
* Для **міста** (номінальна) коректна лише **мода** (найчастіша категорія).
* Для **номера місяця** (порядкова) коректні **мода** та **медіана** (оскільки значення можна впорядкувати).
* Для **кількості опадів** (шкала відношень) коректні **всі три міри**: **мода, медіана та середнє арифметичне**.

## Завдання 2. Варіаційний ряд і частотна таблиця

Побудуємо варіаційний ряд для опадів у Києві, зійдемося на 4 інтервалах і розрахуємо абсолютні, відносні та накопичені частоти.

````
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Дані опадів для Києва (варіант 1)
precip = pd.Series([47, 44, 38, 43, 60, 78, 88, 70, 48, 41, 47, 50])

# 1. Варіаційний ряд (відсортовані значення)
var_series = np.sort(precip.values)
print("1. Варіаційний ряд опадів:", var_series)

# 2. Розбиття на 4 інтервали та побудова частотної таблиці
# Мінімальне значення 38, максимальне 88. Обираємо інтервали: [35-50], [50-65], [65-80], [80-95]
bins = [35, 50, 65, 80, 95]
labels = ["35–50", "50–65", "65–80", "80–95"]

cat = pd.cut(precip, bins=bins, right=True, include_lowest=True, labels=labels)
freq = cat.value_counts().sort_index()
rel = (freq / len(precip) * 100).round(1)
cum = freq.cumsum()

freq_table = pd.DataFrame({
    "абсолютна частота": freq,
    "відносна_%": rel,
    "накопичена частота": cum
})

print("\n2. Частотна таблиця по інтервалах:")
print(freq_table)
````

**Результат виконання:**

```text
1. Варіаційний ряд опадів: [38 41 43 44 47 47 48 50 60 70 78 88]

2. Частотна таблиця по інтервалах:
       абсолютна частота  відносна_%  накопичена частота
35–50                  7        58.3                   7
50–65                  1         8.3                   8
65–80                  3        25.0                  11
80–95                  1         8.3                  12

```

**Відповідь на запитання:**

Частка місяців року, яка має кількість опадів не більшу за межу третього інтервалу (тобто до 80 мм), становить **11 місяців із 12**, що дорівнює накопиченій частоті 11 або **91.6%** від загальної кількості місяців.

## Завдання 3. Гістограма

Будуємо гістограму для кількості опадів із різною кількістю інтервалів (`bins`).

````
# 1. Гістограма із замовчуванням
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.hist(precip, edgecolor="white")
plt.title("За замовчуванням")
plt.xlabel("Опади (мм)")
plt.ylabel("Частота")

# 2. Мала кількість інтервалів (bins=3)
plt.subplot(1, 3, 2)
plt.hist(precip, bins=3, edgecolor="white", color="orange")
plt.title("Мало інтервалів (bins=3)")
plt.xlabel("Опади (мм)")

# 3. Велика кількість інтервалів (bins=10)
plt.subplot(1, 3, 3)
plt.hist(precip, bins=10, edgecolor="white", color="green")
plt.title("Велика кількість інтервалів (bins=10)")
plt.xlabel("Опади (мм)")

plt.tight_layout()
plt.show()

# Орієнтовна кількість інтервалів за правилом сквадратного кореня та Стерджеса для n=12
n = len(precip)
rule_sqrt = np.round(np.sqrt(n))
rule_sturges = np.round(1 + np.log2(n))
print(f"Правило сумісної кількості інтервалів: sqrt(12) ≈ {rule_sqrt}, Стерджес ≈ {rule_sturges}")
````

**Результат виконання:**
*(Графічне зображення трьох гістограм виводиться у середовищі виконання)*

Розрахункові значення:

* За правилом $\sqrt{n} = \sqrt{12} \approx 3.46 \rightarrow 3$ інтервали.
* За правилом Стерджеса $1 + \log_2(12) \approx 4.58 \rightarrow 5$ інтервалів.

**Пояснення:** Гістограма із кількістю інтервалів близько 4–5 найкраще відображає реальний розподіл опадів. Варіант із `bins=3` є занадто узагальненим (надмірне згладжування, втрачаються деталі), а варіант із `bins=10` — занадто подрібненим (більшість стовпчиків мають висоту 1 або 0, що створює хибне враження розірваності даних).

## Завдання 4. Міри положення й квантилі

Обчислюємо основні статистичні характеристики для річних опадів у Києві.

````
mean_val = precip.mean()
median_val = precip.median()
mode_val = precip.mode()[0]
quantiles = precip.quantile([0.25, 0.5, 0.75])

print(f"Середнє арифметичне: {mean_val:.2f}")
print(f"Медіана: {median_val}")
print(f"Мода: {mode_val}")
print("\nКвантилі:")
print(quantiles)
````

**Результат виконання:**

```text
Середнє арифметичне: 54.33
Медіана: 47.5
Мода: 47

Квантилі:
0.25    43.75
0.50    47.50
0.75    67.50
Name: dtype, float64

```

**Пояснення:** Середнє значення (54.33 мм) дещо перевищує медіану (47.5 мм). Така невелика різниця свідчить про помірну правобічну асиметрію розподілу: більшість місяців відносно сухі (зосереджені навколо значень 41–50 мм), проте кілька літніх місяців із високим рівнем опадів (78, 88 мм) зміщують середнє арифметичне вгору.

## Завдання 5. Зв'язок зі шкалою вимірювання

Змінна «кількість опадів (мм)» належить до **шкали відношень**. Саме ця шкала дозволяє коректно виконувати не лише операції порівняння та ранжування, але й повноцінну арифметика (додавання й ділення). Завдяки цьому ми можемо обчислювати **середнє арифметичне**, яке базується на сумуванні всіх значень. Крім того, наявність впорядкованості та абсолютного нуля дозволяє без проблем знаходити медіану і моду.

Натомість змінна «місто» належить до **номінальної шкали**, де можливі лише операції рівності/нерівності (назви категорій). Для номінальних даних неможливо порахувати суму чи середнє (не можна додати «Київ» до «Львова» і поділити на два), тому для них коректною мірою положення є виключно **мода**.

## Письмові відповіді на контрольні запитання

1. **Чи можете обґрунтувати, чому середнє арифметичне некоректне для номінальної змінної «місто», навіть якщо завжди можна закодувати міста числами?**
Номінальні змінні є якісними категоріями. Будь-яке числове кодування (наприклад, Київ = 1, Львів = 2) є довільним і штучним. Розрахунок середнього арифметичного для таких «чисел» дасть беззмістовний результат (наприклад, середнє 1.5 не означає «якесь проміжне місто між Києвом і Львовом»), оскільки над номінальними даними не можна проводити арифметичні операції.
2. **Чи розумієте, чому «номер місяця» — приклад, де класифікація за шкалою вимірювання не однозначна, на відміну від явно порядкової чи явно відносної змінної?**
З одного боку, «номер місяця» має чіткий фіксований порядок (порядкова шкала) і рівні календарні інтервали між послідовними значеннями. Проте через відсутність абсолютного нуля (відсутності точки відліку часу «нульового місяця») він не дотягує до інтервальної чи відносної шкали у сучасному строгом математичному розумінні, що робить його класичним прикладом циклічної або порядкової шкали з рівними інтервалами.
3. **Чи можете пояснити, чому для «кількості опадів» (майже всі 12 значень різні) потрібна саме гістограма з розбиттям на інтервали, а не стовпчикова діаграма за окремими значеннями?**
Стовпчикова діаграма будується для дискретних або категоріальних даних за окремими унікальними значеннями. Оскільки неперервні опади мають майже всі 12 значень різними, стовпчикова діаграма показала б 12 окремих стовпчиків висотою по 1 є одиниці, що не відображає реальної форми розподілу. Гістограма групує дані в інтервали, дозволяючи побачити загальну тенденцію та частоту потрапляння значень у діапазони.
4. **Чи можете пояснити змістовно, що означає накопичена частота на межі вашого третього інтервалу з Завдання 2?**
Накопичена частота на межі третього інтервалу (яка дорівнює 11) означає, що протягом 11 місяців із 12 загальна місячна кількість опадів не перевищувала верхню межу цього інтервалу (80 мм). Тобто це сумарна кількість місяців, що потрапили в перший, другий та третій інтервали включно.
